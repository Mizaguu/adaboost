```python
import pandas as pd
dados = pd.read_csv('C:/Users/Andre/Documents/DidaticaTech/coracao.csv')
dados.head()
```

```python
faltantes = (dados.isnull().sum()/len(dados))*100
faltantes
```

```python
from sklearn.preprocessing import LabelEncoder
def conversao(df):
    le = LabelEncoder()
    for col in df.select_dtypes(include = ['object']):
        df[col] = le.fit_transform(df[col])
    return df
```

```python
conversao(dados)
```

```python
x = dados.drop('HeartDisease', axis=1)
y = dados['HeartDisease']
```

```python
from sklearn.ensemble import AdaBoostClassifier
from sklearn.model_selection import StratifiedKFold,cross_val_score
modelo = AdaBoostClassifier()
skfold = StratifiedKFold()
resultado = cross_val_score(modelo,x,y,cv=skfold,n_jobs=-1)
print(resultado.mean())
```

validando parametros:

```python
import numpy as np
from sklearn.model_selection import GridSearchCV
learning ={'learning_rate': np.array([0.9,0.8,0.7,0.6,0.4,0.3,0.2,0.1,0.01,0.002])}
parametros = GridSearchCV(estimator=modelo, param_grid=learning, cv=3)
parametros.fit(x,y)
print('acuracia:',parametros.best_score_)
```
